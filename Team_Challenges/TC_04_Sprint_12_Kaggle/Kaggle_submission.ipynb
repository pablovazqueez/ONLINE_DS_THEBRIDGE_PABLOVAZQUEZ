## Métrica:
El error cuadrático medio (RMSE, por sus siglas en inglés) es una medida de la desviación estándar de los residuos (errores de predicción). Los residuos representan la diferencia entre los valores observados y los valores predichos por el modelo. El RMSE indica qué tan dispersos están estos errores: cuanto menor es el RMSE, más cercanas están las predicciones a los valores reales. En otras palabras, el RMSE mide qué tan bien se ajusta la línea de regresión a los datos.


$$ RMSE = \sqrt{\frac{1}{n}\Sigma_{i=1}^{n}{\Big(\frac{d_i -f_i}{\sigma_i}\Big)^2}}$$

```python
#Creamos Copias 
X_train_proc = X_train.copy()
X_test_proc = X_test.copy()

#no nos es útil identificador
X_train_proc = X_train_proc.drop(columns=["laptop_ID"])
X_test_proc = X_test_proc.drop(columns=["laptop_ID"])

#convertir Ram de texto a número
X_train_proc["Ram"] = X_train_proc["Ram"].str.replace("GB", "", regex=False).astype(int)
X_test_proc["Ram"] = X_test_proc["Ram"].str.replace("GB", "", regex=False).astype(int)

#convertir Weight de texto a número
X_train_proc["Weight"] = X_train_proc["Weight"].str.replace("kg", "", regex=False).astype(float)
X_test_proc["Weight"] = X_test_proc["Weight"].str.replace("kg", "", regex=False).astype(float)

#comprobación rápida
print(X_train_proc.dtypes)
X_train_proc.head()
```

```python
X_train_proc = pd.get_dummies(X_train_proc, drop_first=False)
X_test_proc = pd.get_dummies(X_test_proc, drop_first=False)

print("Shape antes de alinear:")
print("X_train_proc:", X_train_proc.shape)
print("X_test_proc:", X_test_proc.shape)
```

```python
X_train_proc, X_test_proc = X_train_proc.align(
    X_test_proc,
    join="left",
    axis=1,
    fill_value=0
)

print("Shape después de alinear:")
print("X_train_proc:", X_train_proc.shape)
print("X_test_proc:", X_test_proc.shape)
```

```python
print("¿Quedan columnas object en X_train_proc?", (X_train_proc.dtypes == "object").any())
print("¿Quedan columnas object en X_test_proc?", (X_test_proc.dtypes == "object").any())

print("\nPrimeras filas de X_train_proc:")
X_train_proc.head()
```
